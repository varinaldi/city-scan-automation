```python
def load_raster_gcs(path):
    try:
        with rasterio.open(path) as src:
            data = src.read(1)  # Read first band
            profile = src.profile
            bounds = src.bounds
        return data, profile, bounds
    except Exception as e:
        print(f"Error loading {path}: {e}")
        return None, None, None
```
