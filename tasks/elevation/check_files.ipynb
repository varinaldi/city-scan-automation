```python

import os
from glob import glob
# 
```

```python
scan_id = "2026-02-bangladesh-bogura"
spatial_data = glob(os.path.join('mnt', scan_id, '02-process-output', 'spatial', '*'))
spatial_data.sort()

spatial_data = [os.path.basename(f) for f in spatial_data]
```

```python
import yaml
source = os.path.join('frontend', 'source', 'layers.yml',)


with open(source) as f:
      layers = yaml.safe_load(f)


layers['population']
```

```python
import re

rows = []
for key, val in layers.items():
    pattern = val.get('fuzzy_string')
    if pattern:
        found = any(re.search(pattern, f) for f in spatial_data)
    else:
        found = False
    rows.append({'layer': key, 'pattern': pattern, 'found': found})
```

```python
import pandas as pd
df = pd.DataFrame(rows)
df.sort_values(by='found')
```
